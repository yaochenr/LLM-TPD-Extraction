This notebook add connectivity_key in ground truth set.

```
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
```

```
def process_groundtruth(path):
    df = pd.read_csv(path)
    # Add connectivity_key: first 14 characters of StdInChIKey
    if 'Standard_InChIKey' in df.columns:
        df['Connectivity_Key'] = df['Standard_InChIKey'].str[:14]
    return df
```

```
protac_gt_pth = '/Users/yaochenr/project/data/clean_data_csv/26-04-24-PROTAC-GT.csv'
protac_gt_df = process_groundtruth(protac_gt_pth)
protac_gt_df[['Compound_Name', 'Standard_InChIKey', 'Connectivity_Key']].head()
```
