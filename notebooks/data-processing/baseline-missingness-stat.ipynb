# Missing Data Analysis
Calculate the missingness for all features in both datasets.

```
def calculate_missingness(df):
    """
    Calculate missing percentages for all features.
    """
    missing_pct = (df.isnull().sum() / len(df) * 100).round(4)
    missing_count = df.isnull().sum()
    
    summary = pd.DataFrame({
        'Feature': missing_pct.index,
        'Missing Count': missing_count.values,
        'Missing (%)': missing_pct.values
    }).sort_values(by='Missing (%)', ascending=False).reset_index(drop=True)
    
    return summary
```
